```python
class Solution:
    def reverseWords(self, s: str) -> str:
        s = s.strip()
        words = s.split()       # s.split() returns a new list
        words.reverse()         # Reverse the list in place
        return " ".join(words)

# Test your solution in Jupyter
sol = Solution()
text = "  hello world  "
print(repr(sol.reverseWords(text)))  # Output: 'world hello'
```
